Importing Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Importing Data

```python
main_df = pd.read_csv("https://breathecode.herokuapp.com/asset/internal-link?id=416&path=medical_insurance_cost.csv")
```

Data Check

```python
main_df.info()
main_df.isna().sum()
main_df.duplicated().sum()
main_df.describe()
```

```python
main_df
```

```python
print(main_df.isna().sum())
```

```python
# Check for duplicates
print("Duplicates:", main_df.duplicated().sum())
main_df = main_df.drop_duplicates()
```

```python
# Check data types
print(main_df.dtypes)
```

```python
# Check data descriptions
print(main_df['age'].describe())        # age should be reasonable (0-100ish)
print(main_df['bmi'].describe())        # bmi typically 10-60 range
print(main_df['children'].describe())   # should be non-negative integers
print(main_df['charges'].describe())
```

```python
# Check unique values for categorical variables
print(main_df['sex'].unique())
print(main_df['smoker'].unique())
print(main_df['region'].unique())
```

```python
# Visualize distributions of numerical variables
# Check for outliers visually
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.boxplot(y=main_df['bmi'], ax=axes[0])
sns.boxplot(y=main_df['charges'], ax=axes[1])
sns.boxplot(y=main_df['age'], ax=axes[2])
plt.show()
```
